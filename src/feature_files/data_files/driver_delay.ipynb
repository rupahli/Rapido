```python
import pandas as pd

df_booking  = pd.read_csv(r'C:\Users\KITTU-PALKIN\OneDrive\Documents\Rupahli\GUVI\Projects\Rapido\src\DataCSVFiles\bookings.csv')

df_booking.head()

df_booking.drop(['booking_id', 'actual_ride_time_min', 'drop_location'], axis=1, inplace=True)

df_booking.head()



from sklearn.preprocessing import OrdinalEncoder


encoder = OrdinalEncoder(categories=[[ "Low", #Ordinal
        "Medium",
        "High"]])

df_booking['traffic_level'] = encoder.fit_transform(df_booking[['traffic_level']])

encoder = OrdinalEncoder(categories=[[ "Clear", #Ordinal
        "Rain",
        "Heavy Rain"]])

df_booking['weather_condition'] = encoder.fit_transform(df_booking[['weather_condition']])

df_booking['booking_date'] = pd.to_datetime(df_booking['booking_date'])
df_booking['booking_time'] = pd.to_datetime(df_booking['booking_time'])

df_booking['Fare_per_KM'] = df_booking['booking_value'] / df_booking['ride_distance_km']
df_booking['Fare_per_Min'] = df_booking['booking_value'] / df_booking['estimated_ride_time_min']

df_booking['Long_Distance_Flag'] = df_booking['ride_distance_km'] > 20
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.metrics import f1_score, classification_report, confusion_matrix

X = df_booking.drop(['driver_delay_flag'],axis = 1)
Y = df_booking['driver_delay_flag']

x_train_tuned, x_test_tuned, y_train_tuned, y_test_tuned = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)



rf = RandomForestClassifier(
    random_state=42,
    n_jobs=-1
)

parameter_grid = {
    'n_estimators': [200],
    'max_depth': [10, 14],
    'min_samples_leaf': [5, 10],
    'class_weight': ['balanced_subsample']
}

grid_search = GridSearchCV(
    estimator=rf,
    param_grid=parameter_grid,
    scoring='f1_macro',
    cv=3,
    n_jobs=-1,
    verbose=1
)

grid_search.fit(x_train_tuned, y_train_tuned)

best_rf = grid_search.best_estimator_
tuned_pred = best_rf.predict(x_test_tuned)

print('Best parameters:', grid_search.best_params_)
print('Best cross-validation macro-F1:', grid_search.best_score_)
print('Test macro-F1:', f1_score(y_test_tuned, tuned_pred, average='macro'))
print()
'''print(classification_report(
    y_test_tuned,
    tuned_pred,
    labels=['Completed', 'Cancelled', 'Incomplete'],
    zero_division=0
))
print('Confusion matrix:')
print(confusion_matrix(
    y_test_tuned,
    tuned_pred,
    labels=['Completed', 'Cancelled', 'Incomplete']
))'''

coefficients_df = pd.DataFrame({
    'Feature Name': x_train_tuned.columns,
    'Feature Importance': best_rf.feature_importances_
})

coefficients_df = coefficients_df.sort_values(
    by='Feature Importance',
    ascending=False
)

print('--- Random Forest Feature Importance ---')
print(coefficients_df.to_string(index=False))
```

```python
from sklearn.base import clone
from sklearn.metrics import f1_score

features_to_remove = [
    'booking_year',
    'Long_Distance_Flag',
    'booking_day',
    'is_weekend'
]

X_reduced = X.drop(columns=features_to_remove, errors='ignore')

x_train_reduced, x_test_reduced, y_train_reduced, y_test_reduced = train_test_split(
    X_reduced,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)

reduced_model = clone(best_rf)
reduced_model.fit(x_train_reduced, y_train_reduced)

reduced_pred = reduced_model.predict(x_test_reduced)

print(
    'Reduced test macro-F1:',
    f1_score(y_test_reduced, reduced_pred, average='macro')
)
```
